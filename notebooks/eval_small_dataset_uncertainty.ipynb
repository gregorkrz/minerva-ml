```python
# Generate training cmds
def eval_commands(trainings):
    for loss in trainings.keys():
        for model in trainings[loss].keys():
            for i in range(len(trainings[loss][model])):
                if "omni" in model.lower():
                    print(f"omnilearned evaluate -i {CKPT_DIR}/{trainings[loss][model][i]} --num-workers 2 --batch 100")
                else:
                    print(f"python -m src.scripts.eval --checkpoint {CKPT_DIR}/{trainings[loss][model][i]} --batch_size 1024 --dataset_name minerva_1A --no_amp")
print("")
eval_commands(traning_names_grouped_omni)
```
