```python
import re
import numpy as np
import matplotlib.pyplot as plt
from eval_E_available_plots import plot_rms_iqr_with_uncertainty

# --- Combine all training names across cells into one dict ---
scaling_names = {"Log1p": {}}

for key in ["Transformer 6M", "Transformer 500k", "Transformer 100k"]:
    scaling_names["Log1p"][key] = training_names_grouped["Log1p"][key]

for key in ["OmniS 6M", "OmniS PT 6M", "OmniM PT 6M"]:
    scaling_names["Log1p"][key] = traning_names_grouped_omni["Log1p"][key]

for key in ["OmniS 500k", "OmniS PT 500k", "OmniS 100k", "OmniS PT 100k",
            "OmniS 50k", "OmniS PT 50k"]:
    scaling_names["Log1p"][key] = training_names_grouped_only_PT["Log1p"][key]

# Compute IQR / RMS values for all models via existing function
_, values = plot_rms_iqr_with_uncertainty(
    CKPT_DIR=CKPT_DIR,
    training_names_grouped=scaling_names,
    baseline_run=BASELINE_RUN,
    transform=None,
    return_values=True,
)
plt.close()

# --- IQR vs Number of Training Samples ---
Q3_BIN_IDX = 2

def _n_samples(label):
    m = re.search(r"(\d+(?:\.\d+)?)\s*([kKmM])\b", label)
    if m:
        return float(m.group(1)) * (1e6 if m.group(2).upper() == "M" else 1e3)
    return 0.0

def _method(label):
    return re.sub(r"\s+\d+(?:\.\d+)?[kKmM]$", "", label)

methods: dict[str, list[tuple[float, float, float]]] = {}
for loss in values:
    if loss in ("q3_bin_mids", "baseline"):
        continue
    for cfg, v in values[loss].items():
        methods.setdefault(_method(cfg), []).append(
            (_n_samples(cfg), v["iqr_mean"][Q3_BIN_IDX], v["iqr_std"][Q3_BIN_IDX])
        )

for m in methods:
    methods[m].sort()

method_style = {
    "OmniS":       {"color": "#1b9e77", "linestyle": "-"},
    "OmniS PT":    {"color": "#1b9e77", "linestyle": "--"},
    "OmniM PT":    {"color": "#7570b3", "linestyle": "--"},
    "Transformer": {"color": "#e7298a", "linestyle": "-"},
}

fig, ax = plt.subplots(figsize=(7, 5))
for method, pts in methods.items():
    xs, ys, yerrs = zip(*pts)
    style = method_style.get(method, {"color": "tab:gray", "linestyle": "-"})
    ax.errorbar(xs, ys, yerr=yerrs, marker="o", capsize=4, label=method, **style)

if "baseline" in values:
    ax.axhline(values["baseline"]["iqr"][Q3_BIN_IDX],
               color="black", linestyle=":", lw=1, label="baseline")

ax.set_xscale("log")
ax.set_xlabel("Number of training samples")
ax.set_ylabel("IQR [GeV]")
q3_mid = values["q3_bin_mids"][Q3_BIN_IDX]
ax.set_title(f"IQR vs Training Samples (q$_3$ bin {Q3_BIN_IDX}, mid = {q3_mid:.2f} GeV)")
ax.legend(fontsize=9)
ax.grid(True)
fig.tight_layout()
```
