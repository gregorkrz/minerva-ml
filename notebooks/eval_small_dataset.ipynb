```python
from pathlib import Path
from eval_E_available_plots import load_eval_data, plot_rms_iqr

CKPT_DIR = Path("/global/cfs/cdirs/m3246/gregork/checkpoints")

BASELINE_RUN = "E_avail_LogMSE_20260224_062703"

run_name_list = """SmallDataset_E_avail_LogMSE_1000000_Evts_seed_44_20260225_134252
SmallDataset_E_avail_LogMSE_1000000_Evts_seed_43_20260225_130855
SmallDataset_E_avail_LogMSE_1000000_Evts_seed_42_20260225_130855
SmallDataset_E_avail_LogMSE_500000_Evts_seed_43_20260225_124710
SmallDataset_E_avail_LogMSE_500000_Evts_seed_44_20260225_124710
SmallDataset_E_avail_LogMSE_500000_Evts_seed_42_20260225_123020
SmallDataset_E_avail_LogMSE_100000_Evts_seed_44_20260225_123020
SmallDataset_E_avail_LogMSE_100000_Evts_seed_43_20260225_120449
SmallDataset_E_avail_LogMSE_100000_Evts_seed_42_20260225_120305
SmallDataset_E_avail_LogMSE_4000000_Evts_seed_44_20260227_124529"""

training_names = {"LogMSE": {}}

for line in run_name_list.split("\n"):
    line_strip = line.strip()
    n_events = line_strip.split("_")[4]
    seed = line_strip.split("_")[7]
    training_names["LogMSE"][f"Transformer_{n_events}_{seed}"] = line_strip

training_names["LogMSE"]["Transformer_7M"] = "E_avail_LogMSE_20260224_062703"

data = load_eval_data(CKPT_DIR, training_names, baseline_run=BASELINE_RUN)
```

```python
omnilearned_evals = """SmallDataset_E_avail_LogMSE_PT_dscap1000000_Evts_seed_42_1A_20260225_104358
SmallDataset_E_avail_LogMSE_dscap1000000_Evts_seed_42_1A_20260225_104358
SmallDataset_E_avail_LogMSE_PT_dscap500000_Evts_seed_42_1A_20260225_104355
SmallDataset_E_avail_LogMSE_dscap500000_Evts_seed_42_1A_20260225_104355
SmallDataset_E_avail_LogMSE_dscap100000_Evts_seed_42_1A_20260225_104353
SmallDataset_E_avail_LogMSE_PT_dscap100000_Evts_seed_42_1A_20260225_104353"""

# print eval cmds for those

for line in omnilearned_evals.split("\n"):
    line_strip = line.strip()
    print(f"omnilearned evaluate -i {CKPT_DIR}/{line_strip} --num-workers 2 --batch 100")
```

```python
training_names_omni = {"LogMSE": {}}
for line in omnilearned_evals.split("\n"):
    line_strip = line.strip()
    if not line_strip:
        continue
    parts = line_strip.split("_")
    is_pt = "PT" in parts
    n_events = next(p.replace("dscap", "") for p in parts if p.startswith("dscap"))
    seed = parts[parts.index("seed") + 1]
    model_type = "OmniLearned_Small_Pretrained" if is_pt else "OmniLearned_Small"
    training_names_omni["LogMSE"][f"{model_type}_{n_events}_{seed}"] = line_strip
training_names_omni["LogMSE"]["OmniLearned_Small_Pretrained_7M"] = "E_avail_LogMSE_PT_1A_20260225_102544"
training_names_omni["LogMSE"]["OmniLearned_Small_7M"] = "E_avail_LogMSE_1A_20260225_102540"
data_omni = load_eval_data(CKPT_DIR, training_names_omni, baseline_run=BASELINE_RUN)
```

```python
fig = plot_rms_iqr(
    CKPT_DIR=CKPT_DIR,
    training_names=training_names,
    data=data,
    show_q3_histograms=True,
)
```
