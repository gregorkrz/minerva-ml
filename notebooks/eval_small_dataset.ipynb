```python
from pathlib import Path
from eval_E_available_plots import load_eval_data, plot_rms_iqr, plot_residuals_by_energy

CKPT_DIR = Path("/global/cfs/cdirs/m3246/gregork/checkpoints")

BASELINE_RUN = "E_avail_no_muon_Log1pMSE_20260228_024020"

run_name_list = """SmallDataset_E_avail_Log1PLoss_2000000_Evts_seed_44_20260227_205634
SmallDataset_E_avail_Log1PLoss_2000000_Evts_seed_43_20260227_205634
SmallDataset_E_avail_Log1PLoss_1000000_Evts_seed_44_20260227_204430
SmallDataset_E_avail_Log1PLoss_2000000_Evts_seed_42_20260227_204429
SmallDataset_E_avail_Log1PLoss_1000000_Evts_seed_42_20260227_203125
SmallDataset_E_avail_Log1PLoss_1000000_Evts_seed_43_20260227_203126
SmallDataset_E_avail_Log1PLoss_500000_Evts_seed_44_20260227_200846
SmallDataset_E_avail_Log1PLoss_500000_Evts_seed_43_20260227_200619
SmallDataset_E_avail_Log1PLoss_500000_Evts_seed_42_20260227_194516
SmallDataset_E_avail_Log1PLoss_100000_Evts_seed_44_20260227_194426
SmallDataset_E_avail_Log1PLoss_100000_Evts_seed_43_20260227_191544
SmallDataset_E_avail_Log1PLoss_100000_Evts_seed_42_20260227_191544"""

training_names = {"Log1p": {}}

for line in run_name_list.split("\n"):
    line_strip = line.strip()
    n_events = line_strip.split("_")[4]
    seed = line_strip.split("_")[7]
    training_names["Log1p"][f"Transformer_{n_events}_{seed}"] = line_strip

training_names["Log1p"]["Transformer_6M"] = "E_avail_no_muon_Log1pMSE_20260228_024020"
training_names["Log1p"]["Transformer_6M_seed2"] = "E_avail_Log1PLoss_20260228_155407"



data = load_eval_data(CKPT_DIR, training_names, baseline_run=BASELINE_RUN)
```

```python
def eval_commands(trainings):
    for loss in trainings.keys():
        for model in trainings[loss].keys():
            if "omnilearned" in model.lower():
                print(f"omnilearned evaluate -i {CKPT_DIR}/{trainings[loss][model]} --num-workers 2 --batch 100")
            else:
                print(f"python -m src.scripts.eval --checkpoint {CKPT_DIR}/{trainings[loss][model]} --batch_size 1024 --dataset_name minerva_1A --no_amp")
print("")
eval_commands(training_names)
```
