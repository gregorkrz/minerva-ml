```python
# Parameters: model_dim, depth, n_heads

trainings = """ParamScan_d128_L4_H4_20260301_120625
ParamScan_d128_L6_H8_20260301_120626
ParamScan_d128_L8_H8_20260301_120625
ParamScan_d256_L6_H8_20260301_120624
ParamScan_d128_L4_H16_20260301_120624
ParamScan_d256_L4_H8_20260301_120619
ParamScan_d128_L2_H8_20260301_120619
ParamScan_d128_L4_H8_20260301_120237
ParamScan_d64_L4_H4_20260301_120234"""

# Organize these commands by key like: D128_L4_H4


trainnigs_dict = {"Log1p": {}}
for line in trainings.split("\n"):
    model_name = "_".join(line.split("_")[1:4])
    if model_name not in trainnigs_dict["Log1p"]:
        trainnigs_dict["Log1p"][model_name] = []
    trainnigs_dict["Log1p"][model_name].append(line)
```

```python
CKPT_DIR = "/global/cfs/cdirs/m3246/gregork/checkpoints"
# Generate training cmds
def eval_commands(trainings):
    for loss in trainings.keys():
        for model in trainings[loss].keys():
            if "omni" in model.lower():
                print(f"omnilearned evaluate -i {CKPT_DIR}/{trainings[loss][model][0]} --num-workers 2 --batch 100")
            else:
                print(f"python -m src.scripts.eval --checkpoint {CKPT_DIR}/{trainings[loss][model][0]} --batch_size 1024 --dataset_name minerva_1A --no_amp")
print("")
eval_commands(trainnigs_dict)
```
