```python
from pathlib import Path
from eval_classification_plots import (
    load_results,
    load_truth_and_baselines,
    compute_all_metrics,
    compute_all_metrics_q3,
    compute_signal_baseline,
    plot_cc1pi_vs_pion_kinematics,
    plot_multi_pion_vs_q3,
    plot_binned_by_inttype,
)

CKPT_DIR = Path("/global/cfs/cdirs/m3246/gregork/checkpoints")

training_names = {
    "Transformer": [
        "Pi_Class_v3_20260303_082729",
        "Pi_Class_v3_20260304_145633",
        "Pi_Class_v3_20260304_145825"
    ],
    "OmniLearned_Small": [
        "Train_CC1orNPi_v3_1A_20260303_170126",
        "Train_CC1orNPi_v3_1A_20260304_104006",
        "Train_CC1orNPi_v3_1A_20260304_104020"
    ],
    "OmniLearned_Small_PT": [
        "Train_CC1orNPi_v3_PT_S_1A_20260303_170241",
        "Train_CC1orNPi_v3_PT_S_1A_20260304_104016",
        "Train_CC1orNPi_v3_PT_S_1A_20260304_104006"
    ],
    "OmniLearned_Medium_PT": [
        "Train_CC1orNPi_v3_PT_M_1A_20260304_104018",
        "Train_CC1orNPi_v3_PT_M_1A_20260304_104006"
    ]
}
```

```python
results = load_results(CKPT_DIR, training_names)
data = load_truth_and_baselines(CKPT_DIR, training_names)
```

## CC1π± tagging vs pion kinematics

```python
cc1pi_classes = [0]

metrics_cc1pi = compute_all_metrics(results, data, signal_classes=cc1pi_classes)
baseline_cc1pi = compute_signal_baseline(results, data, signal_classes=cc1pi_classes)

fig = plot_cc1pi_vs_pion_kinematics(
    metrics_cc1pi, data, baseline_cc1pi, uncertainties=True,
)
fig.show()
```

## Multi-pion tagging vs q₃

```python
multi_pi_classes = [0, 1]

metrics_q3 = compute_all_metrics_q3(results, data, signal_classes=multi_pi_classes)
baseline_multi = compute_signal_baseline(results, data, signal_classes=multi_pi_classes)

fig = plot_multi_pion_vs_q3(
    metrics_q3, data, baseline_multi["q3"], uncertainties=True,
)
fig.show()
```

## By interaction type

```python
fig = plot_binned_by_inttype(
    results, data,
    signal_classes=multi_pi_classes,
    x_var="q3",
    xlabel=r"True $q_3$ [GeV]",
    title=r"Multi-pion tagging vs. true $q_3$ — by interaction type",
    uncertainties=True,
)
fig.show()
```

```python
fig = plot_binned_by_inttype(
    results, data,
    signal_classes=cc1pi_classes,
    x_var="pion_E",
    xlabel="Pion energy [GeV]",
    title=r"$CC1\pi^\pm$ tagging vs. pion energy — by interaction type",
    log_x=True,
    uncertainties=True,
)
fig.show()
```

```python
fig = plot_binned_by_inttype(
    results, data,
    signal_classes=cc1pi_classes,
    x_var="pion_theta",
    xlabel="Pion angle [rad]",
    title=r"$CC1\pi^\pm$ tagging vs. pion angle — by interaction type",
    uncertainties=True,
)
fig.show()
```
